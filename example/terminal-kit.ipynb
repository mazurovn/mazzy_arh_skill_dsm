## Два слоя зависимостей
В строке расположен потребитель контракта. Вес симметричной связи учитывается один раз. Числа в ячейках пересчитываются при выполнении.

```python
from collections import Counter
from html import escape
from itertools import combinations
import json
ids = [n["id"] for n in model["nodes"]]
assert len(ids) == len(set(ids))
pos = {v:i for i,v in enumerate(ids)}
def matrix(layer, order):
    index = {n:i for i,n in enumerate(order)}
    rows = [[0]*len(order) for _ in order]
    for e in model["edges"]:
        if e["layer"] != layer: continue
        i,j = index[e["from"]],index[e["to"]]
        assert i != j and not rows[i][j] and type(e["weight"]) == int and e["weight"] > 0
        rows[i][j] = e["weight"]
        if layer == "risk": rows[j][i] = e["weight"]
    return rows
def table(headers, rows):
    try:
        from IPython.display import display, HTML
        tr = lambda cells, tag: "<tr>"+"".join("<{0}>{1}</{0}>".format(tag,escape(str(v))) for v in cells)+"</tr>"
        display(HTML("<table>"+tr(headers,"th")+"".join(tr(r,"td") for r in rows)+"</table>"))
    except ImportError:
        print(headers, *rows, sep="\n")
for layer in ("flow","risk"):
    print(layer)
    table(["ID"] + ids, [[n] + r for n,r in zip(ids, matrix(layer,ids))])
```

## Кластеры и цена изменения
Жадно объединяем группы при положительном приросте модулярности. Сравниваем вес межгрупповых связей и взвешенную длину размещения. Вариант порядка, сохранённый редактором, служит базой сравнения. При новом наборе ID нужно снова экспортировать порядок из редактора.

```python
reference = {"flow":["F04","CFG","CLI","CTX","F01","F06","F02","RUN","F03","LOG","F05","R1","R2","TK"],"risk":["F04","F02","F03","F01","F06","RUN","LOG","F05","CTX","CFG","CLI","R2","R1","TK"]}
def analyze(layer, strong=3):
    weights = Counter()
    for e in model["edges"]:
        if e["layer"] == layer:
            w = strong if layer == "risk" and e["weight"] == 3 else e["weight"]
            weights[tuple(sorted((e["from"],e["to"])))] += w
    total = sum(weights.values())
    deg = Counter()
    for (a,b),w in weights.items(): deg[a] += w; deg[b] += w
    groups = [[n] for n in ids]
    while total:
        membership = {n:i for i,g in enumerate(groups) for n in g}
        mass = [sum(deg[n] for n in g) for g in groups]
        cuts = Counter()
        for (a,b),w in weights.items():
            i,j = sorted([membership[a],membership[b]])
            if i != j: cuts[i,j] += w
        best, pair = 0, None
        for (i,j),w in sorted(cuts.items()):
            gain = w/total - mass[i]*mass[j]/(2*total**2)
            if gain > best + 1e-12: best, pair = gain,(i,j)
        if pair is None: break
        i,j = pair
        groups = [g for k,g in enumerate(groups) if k not in pair]+[groups[i]+groups[j]]
    membership = {n:i for i,g in enumerate(groups) for n in g}
    cross = sum(w for (a,b),w in weights.items() if membership[a]!=membership[b])
    def distance(order):
        if set(order) != set(ids): return None
        p = {v:i for i,v in enumerate(order)}
        return sum(w*abs(p[a]-p[b]) for (a,b),w in weights.items())
    return groups, cross, distance(ids), distance(reference[layer])
for layer in ("flow","risk"):
    groups,cross,before,ref = analyze(layer)
    print(layer, "группы", groups, "межгрупповой вес", cross, "L исходный / базовый порядок", before, ref)
```

## Чувствительность разбиения
Измените вес сильной совместной проверки. Коэффициент Жаккара сравнивает пары узлов, оказавшиеся в одной группе в двух вариантах.

```python
strong_weight = 5
pairs = lambda groups: {tuple(sorted(p)) for group in groups for p in combinations(group, 2)}
a = pairs(analyze("risk")[0])
b = pairs(analyze("risk",strong_weight)[0])
print("Jaccard", len(a & b)/len(a | b) if a | b else 1, "изменившиеся пары", sorted(a ^ b))
```

## План релизов
Оценки указаны в условных единицах работы. Пустая оценка увеличивает число неизвестных. Внешние вызовы показывают точки координации с другими работами.

```python
rows=[]
for r in model["nodes"]:
    if r["kind"] != "release": continue
    assigned = [n for n in model["nodes"] if n.get("release")==r["id"]]
    members={n["id"] for n in assigned}
    known=sum(n["effort"] for n in assigned if n.get("effort") is not None)
    unknown=sum(n.get("effort") is None for n in assigned)
    outside=sum(e["layer"]=="flow" and e["from"] in members and e["to"] not in members for e in model["edges"])
    rows.append([r["id"],len(assigned),known,unknown,r.get("capacity"),outside])
table(["Релиз","Работы","Оценено","Без оценки","Ёмкость","Внешние вызовы"],rows)
```
